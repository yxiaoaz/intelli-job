```python
def get_html(url):
    
    headers = {'User-Agent': USER_AGENT_LIST[1]}
    page = requests.get(url, headers = headers)

    return page
```

### zhilian

```python
page = get_html("https://www.zhaopin.com/jobs")
```

```python
soup = BeautifulSoup(page.text)
print(soup.prettify())
```

```python
zhilian_job_type_items = soup.find_all("a", class_="job-type-item")

print(len(zhilian_job_type_items))
print(zhilian_job_type_items[:5])
```

`et` corresponds to different job types:
- `et = 1`: all
- `et = 2`: full time
- `et = 3`: contract/part time
- `et = 4`: intern
- `et = 5`: graduate

`jl` corresponds to different locations
- `jl=489`: all China

```python
job_item_pages = get_html("https://www.zhaopin.com/sou/jl489/jt14000600130000/p4")
soup = BeautifulSoup(job_item_pages.text)
print(soup.prettify())
```

```python
# get the total number of pages
page_urls = soup.find_all("a", class_="soupager__index")
page_urls[0].text
```

```python
# all job items on a single page

job_items = soup.find_all("a", class_="jobinfo__name")

print(len(job_items))
print(job_items[:5])
```
