### shixiseng

```python
import re
```

```python
shixiseng_intern_page = get_html("https://www.shixiseng.com/interns?type=intern&sortType=zj")
soup = BeautifulSoup(shixiseng_intern_page.text)
print(soup.prettify())
```

```python
# 找到所有 class 为 "intern-wrap interns-point intern-item" 的 div 元素
intern_items = soup.find_all('div', class_='intern-wrap interns-point intern-item')

# 遍历这些元素，提取其中的链接
links = []
for item in intern_items:
    # 在 div 内部查找 a 标签，且 class 包含 "title ellipsis font"
    link = item.find('a', class_='title ellipsis font')
    if link and link.has_attr('href'):
        links.append(link['href'])

# 打印所有找到的链接
for link in links:
    print(link)
```

```python
job_info_page = get_html("https://www.shixiseng.com/intern/inn_uep2vdcv3dtk?pcm=pc_SearchList")
soup = BeautifulSoup(job_info_page.text)
print(soup.prettify())
```

```python

# 1. 提取职位名称
job_name = soup.find('div', class_=r'new_job_name').get_text(strip=True)

# 2. 提取职位发布日期
job_date = soup.find('span', class_='cutom_font').get_text(strip=True)

# 3. 提取薪资、地点、学历要求、实习时长等信息
job_msg = soup.find('div', class_='job_msg')
job_money = job_msg.find('span', class_=re.compile(r'job_money')).get_text(strip=True)
job_location = job_msg.find('span', class_=re.compile(r'job_position')).get_text(strip=True)
job_academic = job_msg.find('span', class_=re.compile(r'job_academic')).get_text(strip=True)
job_week = job_msg.find('span', class_='job_week').get_text(strip=True)
job_duration = job_msg.find('span', class_='job_time').get_text(strip=True)

# 4. 提取职位描述（职责）
job_description = soup.find('div', class_='job_detail').get_text(strip=True)

# 5. 提取截止日期
deadline = soup.find('div', class_='cutom_font', text=lambda t: '截止日期' in str(t)).get_text(strip=True)

# 6. 提取具体工作地点
work_address = soup.find('span', class_='com_position').get_text(strip=True)

# 打印提取的信息
print("职位名称:", job_name)
print("发布日期:", job_date)
print("薪资:", job_money)
print("工作地点:", job_location)
print("学历要求:", job_academic)
print("每周天数:", job_week)
print("实习时长:", job_duration)
print("具体地址:", work_address)
print("截止日期:", deadline)
print("\n职位描述:\n", job_description)
```
